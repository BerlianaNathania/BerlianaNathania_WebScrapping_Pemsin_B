```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.DataFrame([
    [153, 63], [174, 81], [136, 56],
    [186, 91], [128, 47], [136, 57],
    [179, 76], [163, 72], [131, 48]
])

df.columns = ["x", "y"]
```

```python
df.corr()
```

```python
x_train = df["x"].values[:, np.newaxis]
y_train = df["y"].values

x_train
```

```python
lm = LinearRegression()
lm.fit(x_train, y_train)
```

```python
x_test = [[170], [160], [171], [158]]
p = lm.predict(x_test)
p
```

```python
print("Coefficient : ", lm.coef_)
print("Intercept : ", lm.intercept_)
```

```python
pb = lm.predict(x_train)
dfc = pd.DataFrame({'x': df['x'], 'y':pb})
plt.scatter(df['x'], df['y'])
plt.plot(df['x'], dfc['y'], color='magenta', linewidth=1)
plt.xlabel('nilai x')
plt.xlabel('nilai y')
plt.show()
```
